## Variable Assignments & Call Expressions

In Python, there are various types of *objects*, which can be used to store data. To easily access this data, we can give objects names through *variable assignment*. Using variable assignment, we can store values and access them for later use. This is akin to variables we see in math, like $x = 2 * 3 + 5$.

We can use variable assignment to assign a value and use it in another expression. Variables have to start with a letter and can't have any spaces in them. We use a single equal sign (`=`) to assign variables.

```python
num_1 = 56 * 42
```

Here we defined `num_1`, but when we ran the cell, nothing happened. That is because an variable assignment does not warrant any output. If we want to see our newly assigned variable, we will have to *call* for it by typing it in our cell and executing the cell:

```python
num_1
```

Using `num_1`, let's create a new variable called `num_2`

```python
num_2 = num_1 // 7
num_2
```

```python
num_3
```

Sometimes when we are running code, we may skip around or even forget to assign a variable, in which case we get the above error. In general, errors tell you what is wrong and can even give a hint on how to fix it. Let's define `num_3` to resolve this error.

```python
num_3 = num_2 % 36
num_3
```

Note, it is important to understand when to expect an output and when not to. Making a **statement**, such as assigning a variable or importing a library does not give an output. The computer will simply perform the desired task behind the scenes. **Call expressions**, like when we called for `num_1`, `num_2`, and `num_3`, asks the computer to return the named variable, which prints to the console as an output.

Below, we will learn how to call for functions and import other functions from libraries.

## Calling Functions & Importing Libraries

### Calling for functions

Python comes with a lot of built-in functions. A *function* is a set of code that performs a task. To use a function, you must invoke it through a call expression. Just like how we called for a variable above, we can call for functions by typing the function name followed by parentheses. It will look something like:
    `function(...)`

Functions have <u>*arguments*</u> which are used as inputs. Functions take these inputs and perform a task(s). This may result in an output or the task may happen "behind the scenes" (like when we first defined `num_1`).

One of the most basic functions you will use in Python is the `print()` function. As the name suggests, the `print()` function prints an output to the console:

```python
print(5+2)
```

Later, you will see more use of the `print()` function. For now, it's important to understand that the `print()` function allows us to print more than one thing to the console. Calling a variable to print its output only works if that variable is the very last line in the cell. To see the output of multiple variables, we need to use `print()`.

```python
a = 9
b = 11

a
b
```

```python
a = 9
b = 11

print(a)
print(b)
```

Another important function to know in Python is the `help()` function. The `help()` function requires another function as input. In exchange, it provides documentation on how to use that function:

```python
help(print)
```

The `help()` function is useful because it allows you to see how the input function is used, what arguments are required, and other useful information.

### Importing libraries

Sometimes we may need to use things that are not built-in to Python. In these cases, we need to use the `import` statement to import the libraries that we need. Let's import the `math` library, which contains additional math functions that we can use:

```python
import math
```

Now that we've imported the `math` library, we now have access to a bunch of other functions. When we need to use a function from a library, we call the function by first calling the library from which the function derives. Then, we add the function name like so: `library_name.function_name()`.

```python
math.sqrt(4)
```

Above we used the `sqrt()` function in the `math` library to find the square root of the input argument, which was 4.


Functions can also have multiple arguments. The `math` library also has a function called `pow()` which accepts two arguments. The first argument represents the base number. The second argument represents the exponent to which the base is raised. With these input arguments, the `math.pow()` function returns the answer:

```python
math.pow(2,3)
```

For a full list of all the functions you can use from the `math` library, refer to the online documentation:

# Data Types

Python has built-in data types that are used to build data structures, as input for various functions, and many other applications. The major data types we will discuss throughout this program are booleans, strings, integers, and floats.

## Integers

Integers are a numeric data type that consist of whole numbers. Some examples of integers are -14, 9, 20, 0, -7.

Integers can be used to do fast mathematical calculations.

Certain mathematical functions only give integer outputs. For example, the `math.floor()` and `math.ceil()` functions take an expression or variable and rounds it down or up, respectively, to the nearest whole number.

```python
math.floor(7/3)
```

```python
math.ceil(4.82)
```

## Floats

Floats (aka floating point numbers) are a numeric data type that consists of a whole number followed by decimal places. Some examples of floating point numbers are 2.1, -9.453, 3.14159, 0.0

Floats can hold up to 15 significant figures after a decimal point and are used for more accurate calculations.

The `round()` function allows for a float variable to be rounded to a given decimal place:

```python
pi = math.pi
round(pi, 2)
```

Floats and integers can be interchanged using the `float()` and `int()` functions.

```python
print(float(-2))
print(int(pi))
```

Another useful function to know as it pertains to data types is the `type()` function. The type function allows you to determine what data type a variable is.

```python
type(54//9)
```

## Booleans

Booleans are a binary data type consisting of `True` or `False` values. These values equate to the numeric values 1 and 0, respectively. They are often used in comparison operators and control statements (to be discussed in a future lecture).

```python
2 + 3 < 6 - 4
```

You can use the `bool()` function to turn a variable into a boolean. Anything of value will be `True`, anything equal to 0 will be `False`.

```python
bool(0)
```

## Strings

Strings are a series of alphanumeric and punctuation characters. We will spend a lot of time working with strings this summer since they are the main source of data used for NLP (Natural Language Processing), and all of your projects will have an NLP element.

To make a string, place these characters inside a set of single, double, or triple quotations.

```python
phrase = "Is it harder to toot, or to tutor two tooters to toot?"
print(phrase)
```

The number of characters in a string can be determined by using the `len()` function.

```python
len(phrase)
```

We can use multiplication and addition on Strings too. What do you think these will print?

```python
print(2*phrase)
print(phrase+phrase)
```

We see that Python prints the sentence twice, but these sentences run into each other (i.e. there is no space in between).

We have to specifically tell Python to add this space. We can do this by printing the string variables that we want as an expression by using the `+` operator and a space in quotation marks (" "). We call using the `+` operator with Strings *concatenation*.

We can also do this by adding multiple arguments to the `print()` function, separated by a comma.

```python
three = 3
print('My number is', three)
```

```python
print(phrase, phrase)
print(phrase + " " + phrase)
```

Numerical data types can also be converted to strings by using the `str()` function.

```python
five = str(5)
print(five)
five + five
```

Python also has what are called <u>*methods*</u>. A method is like a function in that it performs a task. However, methods are associated with an object, while a function may need an input for it to be used. We can invoke methods by calling for it on an object like so: `object.method_name()`.


Let's use the `.count()` method on `phrase` to count how many times the word "toot" appears:

```python
phrase.count('toot')
```

We can also break a string into pieces using the `.split()` method. The `.split()` method breaks a string anywhere it finds a specified expression.

For example, we can split up `phrase` by the spaces (" ") in it:

```python
phrase.split(" ")
```

This creates a `list` of all the words in `phrases`. We'll learn more about lists later. For now, we can count the number of words by using the above code as an argument for the `len()` function.

```python
len(phrase.split(" "))
```

Finally, we can replace words in a string variable by using the `.replace()` method. With this method, we can specify the expression we want to replace, followed by what we want to replace it with.

```python
phrase.replace("harder", "easier")
```

Documentation for the `.count()`, `.replace()`, and `.split()` methods, as well as other string methods, can be found <a href="https://docs.python.org/3/library/stdtypes.html#textseq">here</a>.

### Escape Sequences

Escape sequences are string modifiers that allow for the use of certain characters that would otherwise be misinterpreted by Python. For example, because strings are created using quotes, the escape sequences `\'` and `\"` allow for the use of quotes as <u>part</u> of a string:

```python
words = "My dad said, \"You\'re going to be okay.\" I looked at him and gave him a hug."
print(words)
```

Though, alternatively, you could just use different types of quotations like so:

```python
words = 'My dad said, "You\'re going to be okay." I looked at him and gave him a hug.'
print(words)
```

Note: You still need to escape the apostrophe in You're as it has no matching end quote.

Other useful escape sequences include `\n` and `\t`. These allow for a new line and tab spacing to be added to a string, respectively.

```python
sentences = 'This is the first sentence \nThis is the second sentence! \tThis is the third sentence?'
print(sentences)
```
