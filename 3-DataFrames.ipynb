## DataFrames are built from *series*

 What is a series?

### Pandas series
#### Pandas defines a basic collection data type called a "series" that is analogous to a one dimentional array in numpy.


````python
Neighborhoods= pd.Series(['Hyde Park','South Lawndale','West Town'])
print(Neighborhoods)
````

We can use series to make DataFrames

In fact, a DataFrame is a sequence of aligned series objects
- “Aligned” means that they share the same index


#### Series are a bit more flexible than arrays.  For example you can specify your own index for a series, and the indexes don't have to be integers.

````python
my_Neighborhoods= pd.Series(['Hyde Park','South Lawndale','West Town'], index = ['a', 'b', 'c'])
print(my_Neighborhoods)
````

We can extract elements from a series either using the numerical index or the self written index.

````python
my_Neighborhoods[1]
````

````python
my_Neighborhoods['b']
````

## Selecting based on label, integer index, and condition (More extraction techniques)

We can extract entire columns or entire rows, but we might want to extract smaller subsets of data. Let's see a few more extraction techniques.


We can use `.loc[]`  if we want to extract via label-based selection that can subset dataframes


We can use `.iloc[]` if we want to extract via integer-based (index based) selection.


Let's take a look at these extraction techniques with the 'small_chicago' dataset so that we can print out exactly what is going on.


````python
small_chicago
````

````python
small_chicago.loc['a','id'] #row label and column label
````

````python
small_chicago.loc[['a', 'c'],'id']
````

````python
small_chicago.loc['a':'c','id':'price']
````

````python
small_chicago
````

````python
small_chicago.iloc[0,0]
````

### Extract by condition

Dataframes can also be selected based on condition using `.loc[]`

That is, based on conditions that equate to `True` Boolean values.

````python
#loc uses the above to keep only those rows that are "True"
small_chicago.loc[small_chicago.neighbourhood == 'Hyde Park']
````

````python
small_chicago.neighbourhood == 'Hyde Park' #we need double = here bc single = is for assignment
````

````python
#as usual, there's alternative syntax to do this
small_chicago['neighbourhood'] == 'Hyde Park'
````

````python
small_chicago.loc[small_chicago.neighbourhood == 'Hyde Park']
````

### Note these two options produce the same result!

You can choose which one you like best.

```python

small_chicago.loc[small_chicago.neighbourhood == 'Hyde Park']

small_chicago[small_chicago.neighbourhood == 'Hyde Park']
```


You can subset on multiple conditions with the use of bitwise operators:

    & for and
    
    | for or


These perform element-wise logical 'and' 'or' operations

Suppose I am interested in...

an Airbnb in Hyde Park **or** an Airbnb less than $80

```python
(small_chicago.neighbourhood == 'Hyde Park') | (small_chicago.price < 80)
```


(It evaluates the truth value of each column of the original DataFrame)

````python
(small_chicago.neighbourhood == 'Hyde Park') | (small_chicago.price < 80)
````

But we want to extract the rows which meet both of the conditions. Hence we denote the DataFrame `small_chicago` that we are extracting from.

````python
small_chicago.loc[(small_chicago.neighbourhood == 'Hyde Park') | (small_chicago.price < 80)]
````

Suppose I am interested in...

an Airbnb in Hyde Park **and** an Airbnb less than $80

````python
small_chicago.loc[(small_chicago.neighbourhood == 'Hyde Park') & (small_chicago.price < 80)]
````

## Some methods help us clean the data!

What are methods again?

*Methods* are particular built-in functions that work on objects in Python. There are specific methods that work for all DataFrame objects!

Methods for DataFrames take the form
```python
df.method()
```

Note, the '.head()' is a method, but '.shape' is an attribute or property. Methods have the () and attributes do not because we can change the argument (or what we input) into the method.

You might have noticed the UK English spelling 'neighbourhood' in the DataFrame we have been working with. We can change that using a method 'rename()' By specifying in the argument (inside the parentheses) a dictionary that maps the current column name to the desired column name.

````python
small_chicago.rename(columns = {"neighbourhood":"neighborhood"})
````

#### Did this change the original DataFrame?!

````python
small_chicago
````

If we want to change the DataFrame, many methods (but not all) have an argument *inplace* that says we want to make a change in place or overwrite the current DataFrame. We can do this using the line below.

````python
small_chicago.rename(columns = {"neighbourhood":"neighborhood"}, inplace = True)
small_chicago
````

Here's a short list of methods that can be used to determine summative information and/or statistics:

`.min()` - determines the minimum value in a dataframe/series

`.max()` - determines the maximum value in a dataframe/series

`.mean()` - determines the mean value of a dataframe/series

`.median()` - determines the median value in a dataframe/series

`.sum()` - determines the sum of the values in a dataframe/series

`.mode()` - determines the mode of a dataframe/series

`.nlargest()` - determines the largest *n* items in a series. (Default = 5)

`.nsmallest()` - determines the smallest *n* items in a series. (Default = 5)

`.count()` - counts all non-null items in a dataframe/series

`.value_counts` - counts the number of times an item occurs in a dataframe/series

`.nunique()` - counts the number of unique items in a dataframe/series

`.unique()` - lists all of the unique items in a dataframe/series

`.describe()` - generates descriptive statistics of a dataframe column, including those that summarize the central tendency, dispersion and shape of a dataset’s distribution, excluding null values (NaN).

We are actually using a default setting here (a setting that is automatically applied unless manually changed) - which is to apply the given function along a column.
````python
small_chicago_clean[['price', 'sqrt price']].apply(np.max, axis = 0)
````

We can change the default if we want to apply a function across rows. This is specified by 'axis = 1', and takes the maximum across rows.

````python
small_chicago_clean[['price', 'sqrt price']].apply(max, axis =1)
````
